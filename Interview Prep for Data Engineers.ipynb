## 📌 Python Problem Set for Senior Data Engineer Interviews

#### This list covers all practical areas required for Tier-1 Senior Data Engineering roles:
#### Core Python, JSON, File Handling, APIs, S3/Boto3, PySpark, Airflow, Algorithms, and Data Engineering logic.

### 1. Core Python Problems Lists, Dicts, Strings
<ul>
<li>Remove duplicates from a list without using set.</li>
<li>Flatten a nested list of arbitrary depth.</li>
<li>Count occurrences of each element in a list.</li>
<li>Reverse words in a sentence without using .split().</li>
<li>Merge two dicts where values are lists and combine them.</li>
<li>Sort a list of dicts by multiple keys.</li>
<li>Filter a list of dicts based on presence of a key.</li>
<li>Group a list of dicts by a key and create a nested structure.</li>
<li>Find common elements between two lists without set.</li>
<li>Convert a list of tuples to a nested dict.</li>
<li>Functions, OOP, Error Handling</li>
<li>Write a function using variable args and kwargs.</li>
<li>Create a DataProcessor class with read, transform, write.</li>
<li>Implement a custom context manager (with).</li>
<li>Write a decorator to measure execution time.</li>
<li>Create custom exception classes for schema and data errors.</li>
</ul>
