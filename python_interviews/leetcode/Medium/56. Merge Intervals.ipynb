```python
'''
Given an array of intervals where intervals[i] = [starti, endi], merge all overlapping intervals, 
and return an array of the non-overlapping intervals that cover all the intervals in the input.

 

Example 1:

Input: intervals = [[1,3],[2,6],[8,10],[15,18]]
Output: [[1,6],[8,10],[15,18]]
Explanation: Since intervals [1,3] and [2,6] overlap, merge them into [1,6].
Example 2:

Input: intervals = [[1,4],[4,5]]
Output: [[1,5]]
Explanation: Intervals [1,4] and [4,5] are considered overlapping.
Example 3:

Input: intervals = [[4,7],[1,4]]
Output: [[1,7]]
Explanation: Intervals [1,4] and [4,7] are considered overlapping.
 

Constraints:

1 <= intervals.length <= 104
intervals[i].length == 2
0 <= starti <= endi <= 104
'''
```

```python
# My Solution 
# if we could avoid sort it could have been best solution

## Solutuion given by need code is the sort


class Solution:
    def merge(self, intervals: list[list[int]]) -> list[list[int]]:
        result = []
        start_interval,end_interval = -1,-1
        #intervals = sorted(intervals)
        intervals.sort(key = lambda i:i[0])
        for i in range(len(intervals)): 

            if start_interval == -1:
                start_interval,end_interval = intervals[i][0],intervals[i][1]
                continue
                
            if start_interval <=  intervals[i][0] and end_interval >=  intervals[i][0] :
                start_interval = min(start_interval, intervals[i][0])
                end_interval   = max(end_interval, intervals[i][1])
            else:
                result.append([start_interval,end_interval])
                start_interval = intervals[i][0]
                end_interval   = intervals[i][1]
        
        result.append([start_interval,end_interval])
        return result
                
```

```python

# Optmized one same to my soltuin 

class Solution:
    def merge(self, intervals: list[list[int]]) -> list[list[int]]:
        result = []
        #intervals = sorted(intervals)
        intervals.sort(key = lambda i:i[0])
        start_interval,end_interval = intervals[0][0],intervals[0][1]
        
        for start,end in intervals[1:]: 
                
            if start_interval <=  start and end_interval >=  start :
                start_interval = min(start_interval, start)
                end_interval   = max(end_interval, end)
            else:
                result.append([start_interval,end_interval])
                start_interval = start
                end_interval   = end
        
        result.append([start_interval,end_interval])
        return result
```
